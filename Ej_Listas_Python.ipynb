13.

```python
numbers = [3, 6, 9, 12, 15]
nueva_lista = [n * 2 for n in numbers]
print(nueva_lista)
```

**Explicación: **

Se usa una comprensión de lista (list comprehension): [n * 2 for n in numbers]

Significa: "para cada n en numbers, calcula n * 2 y guarda el resultado en una nueva lista".

14.

```python
words = ["python", "list", "exercise", "practice"]
for word in words:
    print(word)
```

**Explicación: **

for word in words: Esto es un bucle (ciclo) que va a repetir el código para cada elemento de la lista words

En la primera vuelta, word toma el valor "python"

print(word) : imprime "python"

En la segunda vuelta, word toma el valor "list"

print(word) → imprime "list"

Y así sucesivamente hasta el último elemento

15.

```python
grades = [80, 75, 90, 85, 95]

suma = sum(grades)
cantidad = len(grades)
promedio = suma / cantidad

print(promedio)
```

**Explicación: **

sum(grades) : Suma todos los números de la lista

80 + 75 + 90 + 85 + 95 = 425

len(grades) : Cuenta cuántos elementos tiene la lista

Hay 5 calificaciones

promedio = suma / cantidad : aplica la fórmula:

425 ÷ 5 = 85.0

print(promedio) : Muestra el resultado de la media.
