#Ejercicios de Diccionarios Python

1.

```python
student = {"name": "Daniel", "age": 24, "grade": "Statistics"}
print(student)
```

Explicación:

El comando { }: Define el inicio y fin del diccionario.

El comando :: Une la llave (como "name") con su valor ("Daniel").

Se crea un objeto que contiene múltiples características de una sola persona en una sola variable.

2.

```python
car = {"brand": "Toyota", "model": "Corolla", "year": 2020}
print(car["brand"])
```

Explicación:

El comando [ ]: Se usa para "indexar". Dentro de los corchetes, se pone el nombre de la llave que se quiere consultar.

Es como buscar una palabra en un índice. Busca la etiqueta específica y se devuelva solo lo que está guardado ahí.

3.

```python
car = {"brand": "Toyota", "model": "Corolla", "year": 2020}
print(car["year"])
```

Explicación:

Es igual que el ejercicio anterior solo que en este caso se busca el año.

4.

```python
person = {"name": "Alice", "age": 25}
person["city"] = "Bogota"
print(person)
```

Explicación:

person["city"]: Crea un espacio nuevo que no existía.

=: Guarda el dato "Bogotá" en ese nuevo espacio.

5.

```python
person = {"name": "Alice", "age": 25, "city": "Bogota"}
person["age"] = 26
print(person)
```

Explicación: Al usar una etiqueta que ya existe ("age"), Python sabe que no debe crear nada nuevo, sino reemplazar lo que había antes.

6.

```python
person = {"name": "Alice", "age": 25, "city": "Bogota"}
del person["city"]
print(person)
```

Explicación:

del: Es la "tecla de borrar". Elimina la etiqueta y el contenido de un solo golpe. Si después intentas buscar "city", Python te dirá que no sabe de qué hablas.

7.

```python
fruits = {"apple": 3, "banana": 5, "orange": 2}
print(fruits.keys())
```

Explicación:

Por ejemplo se imagina una canasta de frutas y se quiere hacer una lista de solo los nombres de las frutas que hay, sin que importe cuántas hay de cada una.

Qué hace el comando .keys(): Es indicarle a Python "solo los títulos de las etiquetas".

8.

```python
print(fruits.values())
```

Explicación:

Aca es al contrario se necesitan los números para saber cuántas piezas de fruta se tiene en total, sin importar cuáles son.

Qué hace el comando .values(): Le dice a Python: "Ignora los nombres y lánza solo los datos/números que están guardados".

9.

```python
for fruta, cantidad in fruits.items():
    print(fruta, cantidad)
```

Explicación:

.items(): Se abre el diccionario para ver la pareja (nombre y número) al mismo tiempo.

for: Es el "repartidor" que va pareja por pareja hasta terminar la lista.

10.

```python
grades = {"Ana": 85, "Luis": 90, "Maria": 78}
print("Luis" in grades)
```

Explicación:

in: Es un detector. Revisa rápidamente todas las llaves del diccionario y da una respuesta directa de sí o no.

Explicación:

in: Es un detector. Revisa rápidamente todas las llaves del diccionario y da una respuesta directa de sí o no.

12.

```python
numbers = {"a": 10, "b": 20, "c": 30}
total = sum(numbers.values())
print(total)
```

Explicación: sum(): Toma todos los números que saca .values() y los suma automáticamente. Esto ahorra tener que sumar uno por uno.

13.

```python
inventory = {"pens": 10, "notebooks": 5, "erasers": 8}
print(len(inventory))
```

Explicación:

len(): Viene de length (longitud). Cuenta cuántas "parejas" hay dentro del diccionario.

14.

```python
countries = {"Colombia": "Bogota", "France": "Paris", "Japan": "Tokyo"}
for pais, capital in countries.items():
    print(f"La capital de {pais} es {capital}")
```

Explicación:

.items(): Es el que permite sacar la "pareja completa" (llave y valor) para que el ciclo for los pueda mostrar juntos.

15.

```python
temperatures = {"Monday": 22, "Tuesday": 24, "Wednesday": 21, "Thursday": 23}
print(max(temperatures.values()))
```

Explicación:

max(): Revisa todos los números que le entrega .values() y devuelve únicamente el mayor de todos.
