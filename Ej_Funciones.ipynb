8.

```python
def find_max(a, b):
    if a > b:
        return a
    else:
        return b
resultado = find_max(15, 20)
print(resultado)
```

Explicación:

if a > b:: Es una pregunta. La función dice: "¿Es el primer número más grande que el segundo?".

return a: Si la respuesta es "Sí", la función agarra el primer número lo devuelve.

else:: Significa "De lo contrario".

return b: Si el primero no fue el más grande, entonces por lógica el segundo es el ganador, y devuelve el valor de b.

15.

```python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    else:
        resultado = 1
        for i in range(1, n + 1):
            resultado = resultado * i
        return resultado

print(factorial(5))
```

Explicación:

El factorial es multiplicar un número por todos los que tiene debajo hasta llegar al 1.

El comando for: Es como una bola de nieve que va rodando y multiplicando cada número nuevo por el acumulado que ya tenía.
